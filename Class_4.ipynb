```python
# Exercise 1 — NumPy\n
import numpy as np

# Create array with 20 random integers between 1 and 100
arr = np.random.randint(1, 101, 20)

# 1. Print sum, min, and max
print("Array:", arr)
print("Sum:", np.sum(arr))
print("Min:", np.min(arr))
print("Max:", np.max(arr))

# 2. Create a second array of the same size
arr2 = np.random.randint(1, 101, 20)

# Element-wise multiplication
result = arr * arr2

print("\nSecond Array:", arr2)
print("Element-wise Multiplication:", result)
```

```python
#Exercise 2 — Pandas\n
import pandas as pd
import numpy as np

# Create a 5×3 DataFrame of random integers (1–100)
df = pd.DataFrame(
    np.random.randint(1, 101, size=(5, 3)),
    columns=['A', 'B', 'C']
)

print("DataFrame:")
print(df)

# Sum of each column
print("\nColumn Sums:")
print(df.sum())

# Mean of each column
print("\nColumn Means:")
print(df.mean())
```
